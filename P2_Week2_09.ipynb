```python
import cv2 as cv
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (12, 5)
def pasang_label(img):
    hasil = img.copy()
    h, w = hasil.shape[:2]
    bar_h = h // 12
    skala = h / 1000
    tebal = max(1, int(skala * 2))
    cv.rectangle(hasil, (0, h - bar_h), (w, h), (0, 0, 0), -1)
    cv.putText(hasil, 'Dokumentasi PCVK - 2026', (int(w * 0.02), h - int(bar_h * 0.3)),
               cv.FONT_HERSHEY_SIMPLEX, skala, (255, 255, 255), tebal, cv.LINE_AA)
    return hasil
img1 = cv.imread('rb1.png')
img2 = cv.imread('yusuf2.jpg')
hasil1 = cv.cvtColor(pasang_label(img1), cv.COLOR_BGR2RGB)
hasil2 = cv.cvtColor(pasang_label(img2), cv.COLOR_BGR2RGB)
fig, ax = plt.subplots(1, 2)
ax[0].imshow(hasil1)
ax[0].set_title(f'Hasil Foto 1 ({img1.shape[1]}x{img1.shape[0]} px)')
ax[0].axis('off')
ax[1].imshow(hasil2)
ax[1].set_title(f'Hasil Foto 2 ({img2.shape[1]}x{img2.shape[0]} px)')
ax[1].axis('off')
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (6, 6)
img = cv.imread('bola.jpg')
h, w = img.shape[:2]
sisi = max(h, w)
kanvas = np.full((sisi, sisi, 3), 255, dtype=np.uint8)
y_awal = (sisi - h) // 2
x_awal = (sisi - w) // 2
kanvas[y_awal : y_awal + h, x_awal : x_awal + w] = img
skala = sisi / 1000
tebal = max(1, int(skala * 2))
cv.putText(
    kanvas,
    'Toko Yusuf',
    (int(sisi * 0.03), int(sisi * 0.08)),
    cv.FONT_HERSHEY_SIMPLEX,
    skala,
    (0, 0, 255),
    tebal,
    cv.LINE_AA
)
kanvas_rgb = cv.cvtColor(kanvas, cv.COLOR_BGR2RGB)
plt.imshow(kanvas_rgb)
plt.axis('off')
plt.show()
```
